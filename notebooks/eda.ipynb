```python
import os

# Print current location
print("Current working directory:", os.getcwd())

# Force go to project root
project_root = r"C:\Users\pc\credit-risk-model"
os.chdir(project_root)
print("Changed to:", os.getcwd())

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
%matplotlib inline

df = pd.read_csv('data/raw/training.csv')

print("\n✅ SUCCESS! Dataset loaded")
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
df.head()
```

```python
# Data Overview
print("=== Basic Information ===")
print(df.info())

print("\n=== Missing Values ===")
missing = df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No missing values!")

print("\n=== Summary Statistics ===")
display(df.describe())

print("\n=== First 5 Rows ===")
display(df.head())
```

```python
# Numerical columns
num_cols = ['Amount', 'Value', 'PricingStrategy', 'FraudResult']

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
for i, col in enumerate(num_cols):
    sns.histplot(df[col], kde=True, ax=axes[i//2, i%2])
    axes[i//2, i%2].set_title(f'Distribution of {col}')

plt.tight_layout()
plt.show()
```

```python
cat_cols = ['ProductCategory', 'ChannelId', 'PricingStrategy']

for col in cat_cols:
    print(f"\n=== Top 10 {col} ===")
    print(df[col].value_counts().head(10))
    
    plt.figure(figsize=(10, 5))
    df[col].value_counts().head(10).plot(kind='bar')
    plt.title(f'Top 10 {col}')
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Correlation between numerical features
plt.figure(figsize=(10, 8))
corr = df[['Amount', 'Value', 'PricingStrategy', 'FraudResult']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.show()
```

```python
fraud_rate = df['FraudResult'].mean() * 100
print(f"Fraud Rate: {fraud_rate:.4f}%")
print(f"Number of Fraud cases: {df['FraudResult'].sum()}")
```

### Top 5 Key Insights from EDA

1. **Highly Imbalanced Fraud**: Only ~0.2% of transactions are flagged as fraud. This is typical in fraud detection and will require careful handling during modeling (e.g., class weighting or SMOTE).

2. **Dominant Product Category**: "airtime" and "financial_services" make up the vast majority of transactions, indicating these are the main use cases on the Xente platform.

3. **Transaction Amounts**: Most transactions are small (median around 1,000 - 2,000 UGX), with some very large outliers (utility bills, transport).

4. **Channel Usage**: ChannelId_3 (likely mobile/web) dominates, showing strong mobile money usage in this market.

5. **RFM Potential**: There is significant variation in customer activity (some customers have many transactions, others very few) — this supports using RFM features as a strong proxy for credit risk in Task 4.

**Next Steps**: These insights will guide feature engineering (especially aggregation per CustomerId) and proxy target creation using RFM clustering.
